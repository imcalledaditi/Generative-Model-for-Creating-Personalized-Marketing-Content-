```python
# Step 1: Install Required Libraries
!pip install kaggle nltk

# Step 2: Upload Your Kaggle API Key
from google.colab import files
files.upload()  # Upload your kaggle.json file

# Move the kaggle.json file to the appropriate directory
!mkdir -p ~/.kaggle
!cp kaggle.json ~/.kaggle/
!chmod 600 ~/.kaggle/kaggle.json

# Step 3: Download the Enron Email Dataset from Kaggle
!kaggle datasets download -d wcukierski/enron-email-dataset
!unzip enron-email-dataset.zip

# Step 4: Import Required Libraries
import pandas as pd
import re
import nltk
from nltk import word_tokenize
from collections import defaultdict
import random

# Download NLTK resources
nltk.download('punkt_tab')

# Step 5: Load and Preprocess the Dataset
def clean_email_text(email):

    # Remove email headers (everything before the first blank line)
    email = re.split(r'\n\s*\n', email, maxsplit=1)[-1]

    # Remove URLs, email addresses, and special characters
    email = re.sub(r'http\S+|www\S+', '', email)  # Remove URLs
    email = re.sub(r'\S+@\S+', '', email)        # Remove email addresses
    email = re.sub(r'[^\w\s]', '', email)        # Remove special characters

    # Convert to lowercase and strip extra whitespace
    email = email.lower().strip()

    return email

# Load the dataset into a DataFrame
emails_df = pd.read_csv('emails.csv')  # Adjust the filename if necessary

# Limit the dataset to the first 2,500 emails
emails_df = emails_df.head(2500)

# Display the first few rows of the dataset
print("Dataset Preview:")
print(emails_df.head())

# Extract and clean the email body content
emails_text = emails_df['message'].apply(clean_email_text).tolist()

# Tokenize the text data
tokenized_emails = [word_tokenize(email) for email in emails_text]

# Step 6: Build a Markov Chain Model
# Create a dictionary to store transitions between words
markov_model = defaultdict(list)

# Populate the Markov Chain
for email in tokenized_emails:
    for i in range(len(email) - 1):
        current_word = email[i]
        next_word = email[i + 1]
        markov_model[current_word].append(next_word)

# Step 7: Generate Text Using the Markov Chain
def generate_text(seed_word, num_words=50):

    generated_text = [seed_word]
    current_word = seed_word

    for _ in range(num_words):
        if current_word not in markov_model:
            break
        next_word = random.choice(markov_model[current_word])
        generated_text.append(next_word)
        current_word = next_word

    return ' '.join(generated_text)

# Step 8: Test the Model
# Generate personalized marketing content
seed_text = "offer"
generated_content = generate_text(seed_text, num_words=50)

# Display the generated content
print("\nGenerated Marketing Content:")
print(generated_content)
```
